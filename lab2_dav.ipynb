```python
import numpy as np
import pandas as pd

print("Numpy version: ",np.__version__)
print("Pandas version: ",pd.__version__)
```

```python
marks=np.array([78,85,92,67,88])

print("Marks: ",marks)
print("Type: ",type(marks))
```

Array properties

```python
print("Number of dimension: ",marks.ndim)
print("Shape: ",marks.shape)
print("Number of elements: ",marks.size)
```

```python
marks=np.array([78,85,92,67,88])

print("Original: ",marks)
print("Add 5: ",marks + 5)
print("Multiply by 2: ",marks * 2)
print("Divide by 2: ",marks / 2)
print("Square: ",marks ** 2)
```

```python
print("Sum: ",marks.sum())
print("Mean: ",marks.mean()) #sum of all data points divided by total data points
print("Maximum: ",marks.max())
print("Minimum: ",marks.min())
```

Numpy Indexing

```python
print("First mark: ",marks[0])
print("Second mark: ",marks[1])
print("Last mark: ",marks[-1])
```

Numpy Slicing

```python
print("Index 1 to 3: ",marks[1:4])
print("First three: ",marks[:3])
print("From index 2: ",marks[2:])
print("Every second element: ",marks[::2])
```

Two Dimensional Array

```python
student_marks=np.array([[78,85,92],[67,88,76],[90,95,89]])

print(student_marks)
print("Shape: ",student_marks.shape)
```

```python
print("Mean of all values: ",student_marks.mean())
print("Column-wise mean: ",student_marks.mean(axis=0))
print("Row-wose mean: ",student_marks.mean(axis=1))
```
